```python
import pandas as pd
import numpy as np
from factor_analyzer import FactorAnalyzer
from sklearn.preprocessing import StandardScaler
```

```python
file_path = "clean.xlsx"
df = pd.read_excel(file_path)
```

```python
system_items_clean = [
"Q10","Q11","Q12","Q13","Q14",
"Q15","Q16","Q17","Q18",
"Q31","Q32","Q35"
]
```

```python
pressure_items = [
"Q19","Q20","Q21","Q22",
"Q23","Q24","Q25","Q36"
]
```

```python
scaler = StandardScaler()

X_system = scaler.fit_transform(df[system_items_clean])
X_pressure = scaler.fit_transform(df[pressure_items])
```

```python
fa_system = FactorAnalyzer(
    n_factors=1,
    rotation="varimax",
    method="principal"
)

fa_system.fit(X_system)

df["F1_clean"] = fa_system.transform(X_system)[:,0]
```

```python
fa_pressure = FactorAnalyzer(
    n_factors=1,
    rotation="varimax",
    method="principal"
)

fa_pressure.fit(X_pressure)

df["F2_clean"] = fa_pressure.transform(X_pressure)[:,0]
```

```python
print(pd.DataFrame(
    fa_system.loadings_,
    index=system_items_clean,
    columns=["SystemFactor"]
))

print(pd.DataFrame(
    fa_pressure.loadings_,
    index=pressure_items,
    columns=["PressureFactor"]
))
```

```python
# 机制回归
import statsmodels.api as sm

X = df[["F1_clean","F2_clean"]]
X = sm.add_constant(X)

y = df["Q38"]   # 满意度

model = sm.OLS(y, X).fit()
print(model.summary())
```
